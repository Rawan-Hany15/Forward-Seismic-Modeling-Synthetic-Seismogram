```python
import numpy as np
import matplotlib.pyplot as plt
import lasio

# 1. توليد بيانات اصطناعية مبنية على أسس فيزيائية
np.random.seed(42) # لضمان ثبات النتائج
depths = np.linspace(1000, 1050, 100) # 100 نقطة لعمق البئر

# محاكاة تغير طباقي جيولوجي منظم
base_velocity = 3000 + 1500 * np.sin(depths / 10) + np.random.normal(0, 100, len(depths))

# تحويل السرعة إلى Sonic Log (DT) بالوحدات الحقلية (us/ft)
dt = 304800 / base_velocity

# تطبيق معادلة جاردنر (Gardner's Equation) لحساب الكثافة
v_ft_s = base_velocity * 3.28084
rhob = 0.23 * (v_ft_s ** 0.25) + np.random.normal(0, 0.02, len(depths))

# 2. تفعيل مكتبة lasio وتصدير البيانات بالطريقة الأكثر استقراراً (Standard Syntax)
las = lasio.LASFile()
las.well['WELL'] = lasio.HeaderItem('WELL', value='Synthetic-Well-01')
las.well['STRT'] = lasio.HeaderItem('STRT', value=depths[0], unit='m')
las.well['STOP'] = lasio.HeaderItem('STOP', value=depths[-1], unit='m')
las.well['STEP'] = lasio.HeaderItem('STEP', value=depths[1]-depths[0], unit='m')

# إضافة المنحنيات والبيانات كـ Dictionary (أكثر الطرق استقراراً في lasio)
las['DEPT'] = depths
las['DT'] = dt
las['RHOB'] = rhob

# ضبط الوحدات والوصف لكل منحنى بشكل مستقل لتجنب مشاكل الدوال
las.curves['DEPT'].unit = 'm'
las.curves['DT'].unit = 'us/ft'
las.curves['DT'].descr = 'Sonic Travel Time'
las.curves['RHOB'].unit = 'g/cc'
las.curves['RHOB'].descr = 'Bulk Density'

# حفظ الملف
las.write('synthetic_log.las')
print("تم إنشاء وتصدير ملف synthetic_log.las بنجاح دون أي أخطاء!")

# 3. تصميم لوحة رسم البئر الاحترافية الجيوفيزيائية
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(8, 9), sharey=True)

# المسار الأول: Sonic Log (DT)
ax[0].plot(dt, depths, color='darkred', lw=1.2, label='DT')
ax[0].set_title('Track 1: Sonic Log', fontsize=11, fontweight='bold', pad=15)
ax[0].set_xlabel('DT (us/ft)', fontweight='bold')
ax[0].set_ylabel('Depth (m)', fontsize=12, fontweight='bold')
ax[0].grid(True, linestyle=':', alpha=0.6)
ax[0].set_xlim(40, 140)
ax[0].invert_yaxis() # عكس محور العمق لأسفل

# المسار الثاني: Density Log (RHOB)
ax[1].plot(rhob, depths, color='darkblue', lw=1.2, label='RHOB')
ax[1].set_title('Track 2: Density Log', fontsize=11, fontweight='bold', pad=15)
ax[1].set_xlabel('RHOB (g/cc)', fontweight='bold')
ax[1].grid(True, linestyle=':', alpha=0.6)
ax[1].set_xlim(2.0, 2.8)

# تحسين المظهر العام للوحة
plt.suptitle('Advanced Geophysical Synthetic Log Profile\n(Gardner\'s Relation Applied)',
             fontsize=13, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import lasio

# 1. توليد البيانات الافتراضية (نفس الأساس العلمي السابق)
np.random.seed(42)
depths = np.linspace(1000, 1050, 200) # زيادة النقاط لـ 200 للحصول على جودة سيزمية أعلى

base_velocity = 3000 + 1500 * np.sin(depths / 10) + np.random.normal(0, 80, len(depths))
dt = 304800 / base_velocity
v_ft_s = base_velocity * 3.28084
rhob = 0.23 * (v_ft_s ** 0.25) + np.random.normal(0, 0.01, len(depths))

# ----------------- الخطوة الجديدة -----------------
# 2. الحسابات السيزمية (Seismic Synthetics)

# أ. حساب الأكوستيك إمبيدنس (AI)
ai = base_velocity * (rhob * 1000) # تحويل الكثافة لكيلوجرام/متر مكعب

# ب. حساب معاملات الانعكاس (RC) عند الحدود الفاصلة
rc = np.zeros(len(depths))
rc[1:] = (ai[1:] - ai[:-1]) / (ai[1:] + ai[:-1])

# ج. توليد دالة الموجة السيزمية (Ricker Wavelet)
def ricker_wavelet(f, length=0.08, dt_seismic=0.001):
    t = np.arange(-length/2, length/2, dt_seismic)
    pi2_f2_t2 = (np.pi ** 2) * (f ** 2) * (t ** 2)
    wavelet = (1 - 2 * pi2_f2_t2) * np.exp(-pi2_f2_t2)
    return wavelet

wavelet = ricker_wavelet(f=30, length=0.06) # موجة بتردد 30 هرتز

# د. عمل الـ Convolution لإنتاج الـ Seismic Trace
synthetic_trace = np.convolve(rc, wavelet, mode='same')
# --------------------------------------------------

# 3. تصميم لوحة الرسم الجيوفيزيائية المتكاملة (3 المسارات القياسية)
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(11, 9), sharey=True)

# Track 1: Sonic Log (DT)
ax[0].plot(dt, depths, color='darkred', lw=1.2)
ax[0].set_title('Track 1: Sonic Log', fontsize=10, fontweight='bold', pad=15)
ax[0].set_xlabel('DT (us/ft)', fontweight='bold')
ax[0].set_ylabel('Depth (m)', fontsize=12, fontweight='bold')
ax[0].grid(True, linestyle=':', alpha=0.6)
ax[0].set_xlim(40, 140)
ax[0].invert_yaxis() # عكس العمق لتبدأ من السطح لأسفل

# Track 2: Density Log (RHOB)
ax[1].plot(rhob, depths, color='darkblue', lw=1.2)
ax[1].set_title('Track 2: Density Log', fontsize=10, fontweight='bold', pad=15)
ax[1].set_xlabel('RHOB (g/cc)', fontweight='bold')
ax[1].grid(True, linestyle=':', alpha=0.6)
ax[1].set_xlim(2.0, 2.8)

# Track 3: Synthetic Seismic Trace (المقطع السيزمي الناتجة)
ax[2].plot(synthetic_trace, depths, color='black', lw=1.5, label='Trace')
# تظليل النبضات الموجبة باللون الأسود للحصول على الشكل السيزمي التقليدي (Wiggle Fill)
ax[2].fill_betweenx(depths, 0, synthetic_trace, where=(synthetic_trace >= 0), color='black', alpha=0.7)
ax[2].set_title('Track 3: Seismic Trace', fontsize=10, fontweight='bold', pad=15)
ax[2].set_xlabel('Amplitude', fontweight='bold')
ax[2].grid(True, linestyle=':', alpha=0.6)
ax[2].set_xlim(-0.1, 0.1)

plt.suptitle('Full Geophysical Simulation Portfolio\nFrom Well Logs to Synthetic Seismic Trace',
             fontsize=13, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```
